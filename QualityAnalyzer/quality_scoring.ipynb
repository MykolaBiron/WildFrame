## Calculate the quality of the image by measuring sharpness and object confidence (YOLO)

```python
import os
import cv2
```

```python
# Get sharpness score by computing laplasian variance

def get_sharpness_score(image):
    gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
    score = cv2.Laplacian(gray_image, cv2.CV_64F).var() 
    # 0-50 Blurry image, 50-300 Normal sharpness 300+ Sharp 
    return score

# Test get_sharpness_score on two sample images
sample_image1 = cv2.imread("images/sample_image1.png")
sample_image2 = cv2.imread("images/sample_image2.png")

print(f"Sharp image score: {get_sharpness_score(sample_image1):.3f}")
print(f"Blurry image score: {get_sharpness_score(sample_image2):.3f}")
```
